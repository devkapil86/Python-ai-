## os.path.isfile() -- > This is check the whether is current path is dir or not 
## os.path.isdir()-- > THis is check the whether is current path is dir of not 
## os.path.abspath()--> this is provide the complete path of the abousute path of the file or dir 


```python
path="New_folder"

if os.path.isfile(os.path.abspath(path)): 
    print(" this is file ")
elif os.path.isdir(os.path.abspath(path)):
    print(" this is folder")
else: 
    print(" this is not file or folder")              
```
