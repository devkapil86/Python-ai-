## Nested Dictonary 

Will be used for keeping the a dictonray inside the another dictonary 

I want to store multiple student information liek their name , makrs etc 

```python
nested_dict={"std1":{"name":"kapil","mark(s":90,"roll_no":1},
             "std2":{"name":"raj","mark(s":80,"roll_no":2},
             "std3":{"name":"amit","mark(s":70,"roll_no":3}}
```

```python
nested_dict
```

## iterate the nested dictonary 


```python
for i in nested_dict.items():
    print(i)
```

```python
for outer_key, outer_value in nested_dict.items():
    print(f"outer key is:{outer_key} and outer value is :{outer_value}")
    for key_1,value_1 in outer_value.items():
        print(f"Inner key is :{key_1} and inner values is :{value_1}")
```
