## Tuples 


Tuples are the second type of data structure which can store the data in a orderes way just like list but tuples are not mutable 

Tuples will be created inside these bracket()

```python
a=tuple()
```

```python
print(type(a))
```

```python
tuple_values=(1,2,3,4,5,6,7,8,9)
print(type(tuple_values))
print(tuple_values)
```

Tuples are like list only which can store the data inside it in a way but are going to inmutable 

```python
tuple_values[1]
```

tuples are also capable of storing the data any types of data type

```python
tuple_data=(1,2,3,4,5,6,"amit",2.3,True)
print(tuple_data)
```

tuples are going to be ordered or are going to be indexes 

```python
tuple_data[1:5]
tuple_data[::-1]
```

```python
tuple_data.index("amit")
```

## packing and unpacking of tuples 

```python
a=1,2,3,4,5,6,7,8,9
print(type(a))
print(a)
```

```python
b=1,2,3,"kapil",3.4,True
print(type(b))
print(b)
```

```python
## unpacking of tuple
a,b,c,d,e=(1,2,34,5,4)
print(a)
print(b)
print(c)
print(d)
print(e)
```

## Unpacking with the a * 

(1,2,3,4,)

```python
first_v,*middle_v,final_v=(1,2,3,4,5,5,6,7)
```

```python
print(first_v)
print(middle_v)
print(final_v)
```
