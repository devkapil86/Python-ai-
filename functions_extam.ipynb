## to create a function that converts temprarute from C* to F* and F* to C*

## 1 deg c = 9/5 +32
## 1 Faharenhite c = 5/9 -32 




```python
def temprature_convert( tem_value , unit): 
    if unit=="C": 
        temperature=tem_value * 9/5 +32
        print(f"the tempreture for {tem_value} deg cal in F : ", str(temperature) + "F")
    if unit=='F': 
        temperature=round((tem_value-32 * 5/9),2) 
        print(f"the tempreture for {tem_value} deg F  in F : ", str(temperature) + "C")
           
```

```python
temprature_convert(24,"C")
temprature_convert(80,"F")
```

```python
def even_function(value): 
    if value%2==0: 
        return True
```
