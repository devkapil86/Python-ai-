## Reading the total number of line, words , total characters from the a file ---> Readlines()

## Read line function is used to read the line in the way 
fist_line 
second line 
third line 
.....

Etc

```python
def count_line(file): 
    with open(file,'r') as files: 
        lines=files.readlines()

        total_line=len(lines)
        total_words=sum(len(line.split()) for line in lines)
        total_char=sum(len(line) for line in lines )

        return total_char, total_line, total_words

file_path = 'totaldata.txt'
total_char, total_line, total_words = count_line(file_path)
print(total_char, total_line, total_words)
```
